```python
def lcs(x,y):
    m, n = len(x)+1, len(y)+1
    l = [[0 for j in range(n)] for i in range(m)]
    arrow = [['' for j in range(n)] for i in range(m)]

    for i in range(1, m):
        for j in range(1, n):
            if x[i-1] == y[j-1]:
                l[i][j] = l[i-1][j-1]+1
                arrow[i][j] = 'c' #'↖'
            else:
                if l[i-1][j] >= l[i][j-1]:
                    l[i][j] = l[i-1][j]
                    arrow[i][j] = 'u' # '↑'   
                else:
                    l[i][j] = l[i][j-1]
                    arrow[i][j] = 'l' # '←'

    return l, arrow


def print_lcs(arrow, x, i, j):
    if i == 0 or j ==0:
        return
    if arrow [i][j] == 'c':
        #print(x[i-1])   
        print_lcs(arrow, x, i-1, j-1)
        print(x[i-1])

    elif arrow[i][j] == 'u':
        print_lcs(arrow, x, i-1, j)
    else:
        print_lcs(arrow, x, i, j-1)


x = ['A', 'B', 'C','B', 'D']
y = ['A', 'B', 'D','B', 'B']

l, arrow = lcs(x,y)
print_lcs(arrow, x,len(x),len(y))

    
```

```python
def quick_sort(arr):
    
    if len(arr) <= 1:
        return arr

    pivot = arr[0]
    left = []
    for x in arr:
        if x < pivot:
            left.append(x)

    right = []
    for x in arr:
        if x > pivot:
            right.append(x)
    
    mid = []
    for x in arr:
        if x == pivot:
            mid.append(x)
    
    return quick_sort(left) + mid + quick_sort(right)
```
