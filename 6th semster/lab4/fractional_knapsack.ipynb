```python
def quick_sort(arr):
    
    if len(arr) <= 1:
        return arr

    pivot = arr[0]
    left = []
    right = []
    mid = []
    for x in arr:
        if x < pivot:
            left.append(x)

        elif x > pivot:
            right.append(x)
    
        else:
            mid.append(x)
    
    return quick_sort(left) + mid + quick_sort(right)
```
